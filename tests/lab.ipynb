```python
import requests

VERSION = "v1.1"
SPORT_ID = "1"
```

```python
GAMES_URL = "https://statsapi.mlb.com/api/"+VERSION+"/schedule?sportId="+SPORT_ID
LIVE_GAME_URL = "https://statsapi.mlb.com/api/"+VERSION+"/game/823339/feed/live"
```

```python
res = requests.get(GAMES_URL)
```

```python
games = res.json()['dates'][0]['games']
live_games = [ game for game in games if game['status']['abstractGameState'] == 'Live']
live_games
```

```python
live_games_res = requests.get(LIVE_GAME_URL)
```
